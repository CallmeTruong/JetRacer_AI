# 🏎️ JetRacer Hardware Test Notebook

Simple diagnostic script to test motor steering and throttle control on the **NVIDIA Jetson Nano**.


```python
%load_ext autoreload
%autoreload 2

import sys
import os
import time

try:
    from jetracer.nvidia_racecar import NvidiaRacecar
except ImportError:
    from jetracer_ai.hardware import NvidiaRacecar

car = NvidiaRacecar()
print("✅ NvidiaRacecar HAL Initialized!")
```

```python
# Test Steering Limits (-1.0 to 1.0)
print("Steering Left...")
car.steering = -1.0
time.sleep(1.0)

print("Steering Right...")
car.steering = 1.0
time.sleep(1.0)

print("Centering Steering...")
car.steering = 0.0
```

```python
# Test Motor Throttle (0.15 forward pulse)
print("Forward Pulse...")
car.throttle = 0.15
time.sleep(0.5)

print("Stopping Motor...")
car.throttle = 0.0
```
